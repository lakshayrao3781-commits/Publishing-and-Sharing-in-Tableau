1. What is the key difference between publishing a dashboard on Tableau Public versus Tableau Server/
Cloud in terms of accessibility and data security?

Accessibility & Data Security Difference:

Tableau Public: Completely public and freely accessible to anyone on the internet; data cannot be secured or restricted.  

 Tableau Server/Cloud: Secure, private environment with strict authentication, user roles, and row-level data security.   

2. When publishing a Tableau dashboard, what happens to a live connection compared to an extract?
How does this affect data refresh and performance?

Live Connection: Queries the database directly in real time. Ensures real-time data updates but can cause slower performance depending on database speed.

  Extract: A static snapshot stored in Tableau’s hyper engine. Offers significantly faster query performance but requires scheduled refreshes to update data.   

3. In Tableau Server or Cloud (conceptually), which feature allows administrators to control which users
can view, interact with, or edit a published dashboard?

Permissions & Content Governance (or Role-Based Access Control / Access Rules) set at the project, workbook, or view level.

4. If you want users to view a dashboard but not download the underlying data, which permission setting
would you adjust in Tableau Server?

Deny / uncheck the "Download Summary Data" and "Download Full Data" permissions.

5. Name two methods by which a Tableau dashboard can be embedded into a web page or website.

Using the Tableau JavaScript API / Embedding API v3.  

 Using the standard HTML <iframe> Embed Code copied directly from the dashboard’s Share menu.  

6. List two best practices you should follow when preparing a dashboard for public sharing to ensure data
security and privacy.

Anonymize or aggregate sensitive data (remove PII like names, IDs, addresses).   

Disable the option "Allow viewers to download workbook and its data" in toolbar/publishing settings.   

7. After publishing a dashboard, users report a “Data not found” error.
 What is the most common cause related to data source settings?

The workbook relied on a local file path (e.g., local CSV or Excel on a private computer) instead of an embedded extract or a network-accessible shared database path.

8. When embedding a Tableau dashboard into a secure internal portal, what two additional security
measures should be implemented to ensure only authorized users can access the embedded content?

Single Sign-On (SSO) / SAML / Connected Apps authentication.

  Trusted Authentication (Trusted Tickets / JWT Tokens) to seamlessly verify user identity before rendering.
